```python
import os
import datetime

def julian_to_standard(year, julian_day):
    """Convert a Julian day to a standard date format YYYYMMDD."""
    date = datetime.datetime(year, 1, 1) + datetime.timedelta(julian_day - 1)
    return date.strftime("%Y%m%d")

def process_file(source_file):
    """Reads and processes the file based on Julian dates and stores data into separate files by date in appropriate directories."""
    file_parts = source_file.split('_')
    ftype = '_'.join(file_parts[2:4])
    apu_number = file_parts[1] 
    directory = f"/Users/Indhuja/Desktop/APU/{apu_number}/"  # Give path to store generated files

    # Create directory if it doesn't exist
    if not os.path.exists(directory):
        os.makedirs(directory)

    data_by_date = {}
    with open(source_file, 'r') as file:
        for line in file:
            if line.strip():
                parts = line.split()
                year = int(parts[0])
                julian_day = int(parts[1])
                standard_date = julian_to_standard(year, julian_day)
                if standard_date not in data_by_date:
                    data_by_date[standard_date] = []
                data_by_date[standard_date].append(line)

    # Write the data into separate files within the corresponding directory
    for date, lines in data_by_date.items():
        filename = f"{directory}olympex_{apu_number}_{date}_{ftype}.txt"
        with open(filename, 'w') as file:
            file.writelines(lines)

# To generate apu files for multiple instruments
# source_dir = '/Users/Indhuja/Desktop/APU/olympex/'
# source_files = os.listdir('/Users/Indhuja/Desktop/APU/olympex/')
# index=0
# for file in source_files:
#     fname = os.path.join(source_dir, file)
#     process_file(fname)

source_file = '/Users/Indhuja/Desktop/APU/olympex/olympex_apu08_rainparameter_min_ter.txt' # Source file path
process_file(source_file)
```
