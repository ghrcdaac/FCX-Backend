```python
import xarray as xr
import numpy as np
import matplotlib
matplotlib.use('TkAgg')
import matplotlib.pyplot as plt


file_path = '/Users/Indhuja/Downloads/olympex_d3r_ka_20160115_062855_01.nc'
dataset = xr.open_dataset(file_path)


azimuth = dataset['Azimuth'].values
elevation = dataset['Elevation'].values
reflectivity = dataset['Reflectivity'].values
gates = dataset['Gate'].size


clean_reflectivity = np.where(np.isfinite(reflectivity), reflectivity, np.nan)

# Convert angles from degrees to radians
azimuth_rad = np.deg2rad(azimuth)
elevation_rad = np.deg2rad(elevation)

# Assuming gate spacing is known (e.g., 30 meters)
gate_spacing = 30
ranges = np.linspace(0, gates * gate_spacing, gates)

# Convert from spherical to Cartesian coordinates for 3D plotting
R, AZ = np.meshgrid(ranges, azimuth_rad)
E, _ = np.meshgrid(ranges, elevation_rad)
X = R * np.cos(AZ) * np.cos(E)
Y = R * np.sin(AZ) * np.cos(E)
Z = R * np.sin(E)


fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')


decimation_factor = 5


scatter = ax.scatter(X[::decimation_factor, ::decimation_factor],
                     Y[::decimation_factor, ::decimation_factor],
                     Z[::decimation_factor, ::decimation_factor],
                     c=clean_reflectivity[::decimation_factor, ::decimation_factor],
                     cmap='viridis', marker='o')
ax.set_title('3D Reflectivity Distribution')
ax.set_xlabel('X (m)')
ax.set_ylabel('Y (m)')
ax.set_zlabel('Z (m)')
plt.colorbar(scatter, ax=ax, label='Reflectivity (dBZ)')
plt.show()
```
